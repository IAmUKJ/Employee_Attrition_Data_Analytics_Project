```r
data <- read.csv("WA_Fn-UseC_-HR-Employee-Attrition (1).csv")

head(data)
```

```r
str(data)
summary(data)
```

```r
# Total Employees
nrow(data)
```

```r
# Attrition Count
table(data$Attrition)
```

```r
# Attrition Percentage
prop.table(table(data$Attrition)) * 100
```

```r
# data cleaning
colSums(is.na(data))
```

```r
# Attrition Distribution
ggplot(data, aes(x = Attrition, fill = Attrition)) +
  geom_bar() +
  labs(title = "Employee Attrition Count") +
  theme_minimal()
```

```r
# Attrition by Gender
ggplot(data, aes(x = Gender, fill = Attrition)) +
  geom_bar(position = "dodge") +
  labs(title = "Attrition by Gender") +
  theme_minimal()
```

```r
# Attrition by Department
ggplot(data, aes(x = Department, fill = Attrition)) +
  geom_bar(position = "dodge") +
  labs(title = "Attrition by Department") +
  theme_minimal() +
  coord_flip()
```

```r
# Age vs Attrition
ggplot(data, aes(x = Age, fill = Attrition)) +
  geom_histogram(binwidth = 5, alpha = 0.7) +
  labs(title = "Age Distribution by Attrition") +
  theme_minimal()
```

```r
# Monthly Income vs Attrition
ggplot(data, aes(x = Attrition, y = MonthlyIncome, fill = Attrition)) +
  geom_boxplot() +
  labs(title = "Monthly Income vs Attrition") +
  theme_minimal()
```

```r
# SIMPLE ANALYTICS (INSIGHTS)
```

```r
# Attrition Rate by Department
data %>%
  group_by(Department) %>%
  summarise(
    Total = n(),
    Left = sum(Attrition == "Yes"),
    Attrition_Rate = (Left/Total)*100
  )
```

```r
# Attrition by Job Role
data %>%
  group_by(JobRole) %>%
  summarise(
    Total = n(),
    Left = sum(Attrition == "Yes"),
    Rate = (Left/Total)*100
  ) %>%
  arrange(desc(Rate))
```

```r
# Select numeric columns
num_data <- data[sapply(data, is.numeric)]

# Remove zero-variance columns
num_data <- num_data[, apply(num_data, 2, sd) != 0]

# Compute correlation
corr_matrix <- cor(num_data)

# Plot again
corrplot(corr_matrix, method="color", tl.cex=0.6)
```
